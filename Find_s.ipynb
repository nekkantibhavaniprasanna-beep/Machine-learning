```python
import pandas as pd
data={
    'sky':['sunny','sunny','rainy','sunny'],
    'temp':['warm','warm','cold','cold'],
    'humidity':['normal','high','high','normal'],
    'wind':['weak','strong','weak','strong'],
    'water':['warm','warm','warm','cold'],
}
df=pd.DataFrame(data)
display(df)
```

```python
def find_s(df):
  concepts=df.iloc[:,:-1].values
  target=df.iloc[:,-1].values
  num_attributes=concepts.shape[1]
  hypothesis=[0]*num_attributes

  for index,instance in enumerate(concepts):
    if target[index]=="yes":
      hypothesis=instance.copy()
      break

  for index,instance in enumerate(concepts):
    if target[index]=="yes":
      for x in range(num_attributes):
        if instance[x]==hypothesis[x]:
          continue
        else:
          hypothesis[x]='?'
  return hypothesis

most_specific_hypothesis=find_s(df)
print(most_specific_hypothesis)





```
